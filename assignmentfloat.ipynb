```python
#Calculate the average of 3.14, 2.718, 1.618, 0.577
numbers = [3.14, 2.718, 1.618, 0.577]
average = sum(numbers) / len(numbers)
print(f"Average: {average}")
```

```python
#Convert 98.6 Fahrenheit to Celsius (F = C * 9/5 + 32)
fahrenheit = 98.6
celsius = (fahrenheit - 32) * 5 / 9
print(f"Celsius: {celsius}")
```

```python
# Calculate the compound interest on $1000 at 5.5% for 3 years
principal = 1000
rate = 5.5 / 100
time = 3
amount = principal * (1 + rate) ** time
compound_interest = amount - principal
print(f"Compound Interest: ${compound_interest:.2f}")
print(f"Final Amount: ${amount:.2f}")
```

```python
#Find the hypotenuse of a right triangle with sides 3.5 and 4.2
import math
side1 = 3.5
side2 = 4.2
hypotenuse = math.sqrt(side1 ** 2 + side2 ** 2)
print(f"Hypotenuse: {hypotenuse:.2f}")
```

```python
#Calculate the volume of a sphere with radius 7.8
radius = 7.8
volume = (4 / 3) * math.pi * radius ** 3
print(f"Volume: {volume:.2f}")
```

```python
#Round 3.14159 to 3 decimal places
number = 3.14159
rounded_number = round(number, 3)
print(f"Rounded number: {rounded_number}")
```

```python
#Calculate the percentage: 45 out of 67

percentage = (45 / 67) * 100
print(f"Percentage: {percentage:.2f}%")
```

```python
#Find the square root of 23.456
number = 23.456
square_root = math.sqrt(number)
print(f"Square root: {square_root:.2f}")
```

```python
# Calculate the simple interest: Principal=2500, Rate=6.5%, Time=2.5 years
principal = 2500
rate = 6.5 / 100
time = 2.5
simple_interest = principal * rate * time
print(f"Simple Interest: {simple_interest}")
```

```python
#Convert 45.7 degrees to radians

degrees = 45.7
radians = math.radians(degrees)
print(f"Radians: {radians:.4f}")
```
