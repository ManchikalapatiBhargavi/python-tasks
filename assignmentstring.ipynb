```python
#23: Convert to alternating case

text = "hello world"
result = ""

index = 0
for char in text:
    if char.isalpha():
        if index % 2 == 0:
            result += char.lower()
        else:
            result += char.upper()
        index += 1
    else:
        result += char

print(result)
```

```python
#41: Check balanced parentheses
text = "((()))"
count = 0
balanced = True

for char in text:
    if char == "(":
        count += 1
    elif char == ")":
        count -= 1

    if count < 0:
        balanced = False
        break

if count != 0:
    balanced = False

print(balanced)
```

```python
#42: Convert "hello world" to Morse code


morse = {
    "a": ".-", "b": "-...", "c": "-.-.", "d": "-..",
    "e": ".", "f": "..-.", "g": "--.", "h": "....",
    "i": "..", "j": ".---", "k": "-.-", "l": ".-..",
    "m": "--", "n": "-.", "o": "---", "p": ".--.",
    "q": "--.-", "r": ".-.", "s": "...", "t": "-",
    "u": "..-", "v": "...-", "w": ".--", "x": "-..-",
    "y": "-.--", "z": "--.."
}

text = "hello world"
result = ""

for char in text:
    if char == " ":
        result += " / "
    else:
        result += morse[char] + " "

print(result)
```

```python
#43: Find longest common substring

text1 = "programming"
text2 = "grammar"

longest = ""

for i in range(len(text1)):
    for j in range(i + 1, len(text1) + 1):
        substring = text1[i:j]

        if substring in text2 and len(substring) > len(longest):
            longest = substring
print(longest)         
```

```python
# 46: Convert "hello world" to Pig Latin


text = "hello world"
words = text.split()
result = []

for word in words:
    result.append(word[1:] + word[0] + "ay")

print(" ".join(result))
```

```python
#47: Check valid IPv4 address

ip = "192.168.1.1"
parts = ip.split(".")

valid = len(parts) == 4

if valid:
    for part in parts:
        if not part.isdigit() or not 0 <= int(part) <= 255:
            valid = False
            break

print(valid)
```

```python
#50: Check valid credit card number

card = "4532015112830366"

digits = [int(x) for x in card]
total = 0

for i in range(len(digits) - 1, -1, -1):
    digit = digits[i]

    if (len(digits) - i) % 2 == 0:
        digit *= 2

        if digit > 9:
            digit -= 9

    total += digit

print(total % 10 == 0)
```
